```python
import pandas as pd
import numpy as np

# Menggunakan URL raw agar file CSV dapat dibaca langsung oleh pandas
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Menampilkan 5 data teratas untuk memastikan data termuat
df.head()
```

```python
# Q1: Berapa jumlah baris dan kolom pada dataset titanic? Kolom apa saja yang tersedia?

# Mendapatkan Jumlah baris dan kolom
rows, cols = df.shape
print(f"Jumlah Baris: {rows}")
print(f"Jumlah Kolom: {cols}")

# Menampilkan daftar nama kolom
print("Daftar Kolom:")
print(df.columns.tolist())
```

```python
# Q2: Kolom mana saja yang memiliki missing values? Berapa jumlah dan proporsinya (%) per kolom?

# Menghitung jumlah missing values
missing_count = df.isnull().sum()

# Menghitung proporsi (%)
missing_percentage = (missing_count / len(df)) * 100

# Menggabungkan hasil ke dalam satu DataFrame agar mudah dibaca
missing_data = pd.DataFrame({
    'Jumlah Missing': missing_count,
    'Persentase (%)': missing_percentage
})

print(missing_data[missing_data['Jumlah Missing'] > 0])
```

```python
# Q3: Berapa persen penumpang yang selamat (Survived=1)? Gunakan operasi NumPy atau Pandas.

# Menghitung rata-rata dari kolom Survived (karena 1 = Selamat, 0 = Tidak)
survival_rate = df['Survived'].mean() * 100
print(f"Persentase penumpang selamat: {survival_rate:.2f}%")
```

```python
# Q4: Filter penumpang wanita dari kelas 1 (Pclass=1). Berapa jumlahnya dan berapa rata-rata usianya?

# Melakukan Filtering
female_class1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]

# Menghitung jumlah dan rata-rata usia
count_female_c1 = len(female_class1)
average_age_female_c1 = female_class1['Age'].mean()

print(f"Jumlah penumpang wanita kelas 1: {count_female_c1}")
print(f"Rata-rata usia mereka: {average_age_female_c1:.2f} tahun")
```

```python
# Q5: Gunakan groupby untuk menghitung tingkat keselamatan (mean Survived) per kelas penumpang. Kelas mana yang paling tinggi?

# Menggunakan Groupy untuk melihat mean Survived per Pclass
survival_by_class = df.groupby('Pclass')['Survived'].mean() * 100
print(survival_by_class)
```

```python
# Q6: Buat analisis tambahan berdasarkan variabel lain yang Anda anggap menarik. Sertakan interpretasi singkat.

# Menghitung tingkat keselamatan berdasarkan jenis kelamin
survival_by_gender = df.groupby('Sex')['Survived'].mean() * 100
print(survival_by_gender)
```

Temuan: Terdapat perbedaan yang sangat kontras; wanita memiliki tingkat keselamatan sekitar 74%, sedangkan pria hanya sekitar 18%. Hal ini memperkuat teori kebijakan "wanita dan anak-anak didahulukan" saat evakuasi. Selain itu, menarik untuk melihat korelasi antara tarif tiket (Fare) dengan keselamatan, di mana biasanya tarif yang lebih mahal memberikan akses evakuasi yang lebih cepat.
